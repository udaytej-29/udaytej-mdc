Solutions for the three requested programming problems.

### 1. Numbers divisible by 7 but not a multiple of 3 (2000 to 3200)

```python
nl = []
for x in range(2000, 3201):
    if (x % 7 == 0) and (x % 3 != 0):
        nl.append(str(x))
print(",".join(nl))
```

### 2. Convert a list of characters into a string

```python
chars = ['C', 'O', 'D', 'I', 'N', 'G']
result = "".join(chars).capitalize()
print(result)
```

### 3. Generate a dictionary containing (i, i*i) for a given integral number n

```python
n = 5
ans_dict = {i: i*i for i in range(1, n + 1)}
print(ans_dict)
```
